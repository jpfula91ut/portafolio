**EJERCICIO 5 - Cree un programa que solicite la longitud de los 3 lados de** **un triangulo e imprima si es equilátero(3 lados iguales), isósceles(2 lados** **iguales) o escaleno(todos los lados diferentes)**

```python
lado1 = float(input("Ingrese la longitud del primer lado: "))
lado2 = float(input("Ingrese la longitud del segundo lado: "))
lado3 = float(input("Ingrese la longitud del tercer lado: "))
if lado1 == lado2 == lado3:
    print("El triangulo es equilatero")
elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
    print("El triangulo es isoceles")
elif lado1 != lado2 != lado3:
    print("El triangulo es escaleno")


```

**EJERCICIO 6 -**
**En una empresa trabajan n empleados cuyos sueldos oscilan entre $100 y $500** **realizar un programa que lea los sueldos que cobra cada empleado e** **informe cuántos empleados cobran entre $100 y $300 y cuántos cobran más de** **$300. Además el programa deberá informar el importe que gasta la empresa** **en sueldos al personal**

```python
num_empleados = int(input("Ingrese el número de empleados: "))

#Inicializar variables con un valor de cero
empleados_100_300 = 0
empleados_mas_300 = 0
total_gastado = 0

#Bucle llamando a la variable i mientras este en el rango de numero de empleados
for i in range(num_empleados):
  #Mientras este en bucle solicita un sueldo del empleado ennumerado
    sueldo = float(input(f"Ingrese el sueldo del empleado {i+1}: "))
    #A medida que se cumple suma al total gastado la cantidad de sueldo
    total_gastado += sueldo

    #Condiciones que evalua el sueldo del empleado para añadirlo a un grupo
    if 100 <= sueldo <= 300:
        empleados_100_300 += 1
    elif sueldo > 300:
        empleados_mas_300 += 1

# print(f()) ingresa valores sin formato dentro de un texto
print(f"La empresa gasta un total de ${total_gastado} en sueldos.")
print(f"{empleados_100_300} empleados cobran entre $100 y $300.")
print(f"{empleados_mas_300} empleados cobran más de $300.")
```

EJERCICIO 7 - Una planta que fabrica perfiles de hierro posee un lote de npiezas
Diseñar un programa que pida ingresar por teclado la cantidad de piezas     
a procesar e ingrese la longitud de cada perfil, sabiendo que la pieza cuya longitud este comprendida entre 1.20 y 1.30 son aptas. Imprimir la cantidad de piezas del lote que son aptas

```python
cantidad_piezas = int(input("Ingrese la cantidad de piezas a procesar: "))
piezas_aptos = 0
for i in range(cantidad_piezas):
    longitud = float(input(f"Ingrese la longitud de la pieza {i+1}: "))
    if 1.20 <= longitud <= 1.30:
        piezas_aptos += 1
print(f"La cantidad de piezas aptas es: {piezas_aptos}")

```
